```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
data = pd.read_csv("heart.csv")
```

```python
print(data.head())
```

```python
print(data.columns)
```

```python
x = data.drop("target",axis=1)
y = data["target"]
```

```python
print("X:")
print(x.head())

print("\ny:")
print(y.head())
```

```python
print("Missing values in each column")
print("-------------------------------")
print(data.isnull().sum())
```

```python
print("Number of Duplicate Rows")
print("------------------------")
print(data.duplicated().sum())
```

```python
data=data.drop_duplicates()
print("Duplicates removed successfully")
print("---------------------------------")
print("Number of rows after removing duplicates:", len(data))
```

```python
x = data.drop("target", axis=1)
y = data["target"]
print("Features (X) shape:", x.shape)
print("Target (y) shape:", y.shape)
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y
)
```

```python
print("Training data shape",x_train.shape)
print("Testing data shape", x_test.shape)
```

```python
model = DecisionTreeClassifier(random_state=42)
model.fit(x_train, y_train)
```

```python
train_predictions = model.predict(x_train)

test_predictions = model.predict(x_test)

print("Predictions Generated Successfully")
```

```python
train_accuracy = accuracy_score(y_train, train_predictions)

print("Training Accuracy")
print("-----------------")
print(train_accuracy)
```

```python
test_accuracy = accuracy_score(y_test, test_predictions)

print("Testing Accuracy")
print("----------------")
print(test_accuracy)
```

```python
# Pre-Pruned Decision Tree

model2 = DecisionTreeClassifier(
    max_depth=3,
    random_state=42
)

model2.fit(x_train, y_train)
```

```python
pre_train_predictions = model2.predict(x_train)
pre_test_predictions = model2.predict(x_test)

print("Predictions Generated Successfully")
```

```python
pre_train_accuracy = accuracy_score(y_train, pre_train_predictions)
pre_test_accuracy = accuracy_score(y_test, pre_test_predictions)

print("Pre-Pruned Decision Tree Results")
print("--------------------------------")
print(f"Training Accuracy : {pre_train_accuracy * 100:.2f}%")
print(f"Testing Accuracy  : {pre_test_accuracy * 100:.2f}%")
```

```python
path = model.cost_complexity_pruning_path(x_train, y_train)

ccp_alphas = path.ccp_alphas

print("Number of CCP Alpha Values:", len(ccp_alphas))
print("CCP Alpha Values:")
print(ccp_alphas)
```

```python
train_scores = []
test_scores = []

for alpha in ccp_alphas:
    post_pruned_model = DecisionTreeClassifier(
        random_state=42,
        ccp_alpha=alpha
    )

    post_pruned_model.fit(x_train, y_train)

    train_score = post_pruned_model.score(x_train, y_train)
    test_score = post_pruned_model.score(x_test, y_test)

    train_scores.append(train_score)
    test_scores.append(test_score)
```

```python
print("CCP Alpha | Training Accuracy | Testing Accuracy")
print("-" * 55)

for alpha, train_score, test_score in zip(
    ccp_alphas,
    train_scores,
    test_scores
):
    print(f"{alpha:.6f} | {train_score:.4f}            | {test_score:.4f}")
```

```python
# Post-Pruned Decision Tree

post_pruned_model = DecisionTreeClassifier(
    ccp_alpha=0.026366,
    random_state=42
)

post_pruned_model.fit(x_train, y_train)

print("Post-Pruned Decision Tree Trained Successfully")
```

```python
post_train_predictions = post_pruned_model.predict(x_train)
post_test_predictions = post_pruned_model.predict(x_test)

post_train_accuracy = accuracy_score(
    y_train,
    post_train_predictions
)

post_test_accuracy = accuracy_score(
    y_test,
    post_test_predictions
)

print("Post-Pruned Decision Tree Results")
print("---------------------------------")
print(f"Training Accuracy : {post_train_accuracy * 100:.2f}%")
print(f"Testing Accuracy  : {post_test_accuracy * 100:.2f}%")
```

```python
print("FINAL DECISION TREE COMPARISON")
print("=" * 50)

print("\nUnpruned Decision Tree")
print("-" * 50)
print(f"Training Accuracy : {train_accuracy * 100:.2f}%")
print(f"Testing Accuracy  : {test_accuracy * 100:.2f}%")

print("\nPre-Pruned Decision Tree")
print("-" * 50)
print(f"Training Accuracy : {pre_train_accuracy * 100:.2f}%")
print(f"Testing Accuracy  : {pre_test_accuracy * 100:.2f}%")

print("\nPost-Pruned Decision Tree")
print("-" * 50)
print(f"Training Accuracy : {post_train_accuracy * 100:.2f}%")
print(f"Testing Accuracy  : {post_test_accuracy * 100:.2f}%")
```
