```python
#Activity10:Find total no of passengers, avg age of survivers and non survivers, list the passengers who are older 60yrs, and find the avg fare for each passengers
import pandas as pd
df = pd.read_csv("Titanic-Dataset.csv")
df.head()
```

```python
df.info()
df.describe()
```

```python
total_passengers = df["PassengerId"].count()
print("Total no of passengers:", total_passengers)
```

```python
survivors = df[df["Survived"]==1]
avg_survivor_age = survivors["Age"].mean()
print("Avg age of survivors:", avg_survivor_age)
```

```python
non_survivors = df[df["Survived"]==0]
avg_non_survivor_age = non_survivors["Age"].mean()
print("Avg age of non survivors:", avg_non_survivor_age)
```

```python
older_60 = df[df["Age"]>60]
print("Passengers older than 60 years:")
print(older_60)
```

```python
avg_fare = df["Fare"].mean()
print("Average fare:", avg_fare)
```

```python
#Activity 4: Calculating the weather data of manipal
import pandas as pd
df = pd.read_csv("weatherdata.csv", skiprows=9)
```

```python
print(df.head())
```

```python
print(df.columns.tolist())
```

```python
temp = pd.to_numeric(
    df["Basel Temperature [2 m elevation corrected]"],
    errors = "coerce"
)
```

```python
print("Minimum temperature:", temp.min())
```

```python
print("Maximum temperature:", temp.max())
```

```python
print("Average temperature:", temp.mean())
```

```python
above_35 = temp[temp > 35]

print(above_35)
```

```python
cold = temp[temp <= 35]
heat = temp[temp > 35]

print("Minimum cold:", cold.min())
print("Maximum cold:", cold.max())
print("Minimum heat:", heat.min())
print("Maximum heat:", heat.max())
```
